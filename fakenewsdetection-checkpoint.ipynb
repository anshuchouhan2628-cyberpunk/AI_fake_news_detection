```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt 
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report
import re
import string
```

```python
data_fake=pd.read_csv('Fake.csv')
data_true=pd.read_csv('True.csv')
```

```python
data_true.head()
```

```python
data_fake["class"]=0
data_true["class"]=1
```

```python
data_fake_manual_testing=data_fake.tail(10)
for i in range(23480, 23470,-1):
    data_fake.drop([i], axis=0,inplace = True)

data_true_manual_testing=data_true.tail(10)
for i in range(21416, 21406,-1):
    data_true.drop([i], axis=0,inplace = True)
```

```python
data_fake_manual_testing['class']=0
data_true_manual_testing['class']=1
```

```python
data_fake_manual_testing.head(10)
```

```python
data_true_manual_testing.head(10)
```

```python
data_merge=pd.concat([data_fake,data_true], axis =0)
data_merge.head(10)
```

```python
data_merge.columns
```

```python
data=data_merge.drop(['title','subject','date'],axis=1)
```

```python
data.isnull().sum()
```

```python
data=data.sample(frac=1)
```

```python
data.reset_index(inplace=True)
data.drop(['index'],axis=1,inplace=True)
```

```python
data.columns
```

```python
data.head()
```

```python
def wordopt(text):
    text = text.lower()
    text = re.sub(r'\[.*?\]', '', text)
    text = re.sub(r"\W", " ", text)
    text = re.sub(r'https?://\S+|www\.\S+', '', text)
    text = re.sub(r'<.*?>+', '', text)
    text = re.sub(r'[%s]' % re.escape(string.punctuation), '', text)
    text = re.sub(r'\n', '', text)
    text = re.sub(r'\w*\d\w*', '', text)
    return text
```

```python
data['text'] = data['text'].apply(wordopt)
```

```python
x= data['text']
y=data['class']
```

```python
x_train, x_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorization = TfidfVectorizer()
xv_train = vectorization.fit_transform(x_train)
xv_test = vectorization.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
LR=LogisticRegression()
LR.fit(xv_train,y_train)
```

```python
pred_ir=LR.predict(xv_test)
```

```python
LR.score(xv_test,y_test)
```

```python
print(classification_report(y_test,pred_ir))
```

```python
from sklearn.tree import DecisionTreeClassifier
DT=DecisionTreeClassifier()
DT.fit(xv_train,y_train)
```

```python
pred_dt=DT.predict(xv_test)
```

```python
DT.score(xv_test,y_test)
```

```python
print(classification_report(y_test,pred_dt))
```

```python
from sklearn.ensemble import GradientBoostingClassifier
GB=GradientBoostingClassifier(random_state=0)
GB.fit(xv_train,y_train)
```

```python
predict_gb=GB.predict(xv_test)
```

```python
GB.score(xv_test,y_test)
```

```python
print(classification_report(y_test,predict_gb))
```

```python
from sklearn.ensemble import RandomForestClassifier
RF=RandomForestClassifier(random_state=0)
RF.fit(xv_train,y_train)
```

```python
pred_rf=RF.predict(xv_test)
```

```python
RF.score(xv_test,y_test)
```

```python
print(classification_report(y_test,pred_rf))
```

```python

def output_label(n):
    if n == 0:
        return "Fake News"
    elif n == 1:
        return "Real News"
    else:
        return "Unknown"


def manual_testing(news):
    testing_news = {"text": [news]}
    new_def_test = pd.DataFrame(testing_news)

    # Apply the same cleaning used during training
    new_def_test["text"] = new_def_test["text"].apply(wordopt)

    # Convert text into numerical features
    new_x_test = new_def_test["text"]
    new_xv_test = vectorization.transform(new_x_test)

    # Predict using all four models
    pred_LR = LR.predict(new_xv_test)
    pred_DT = DT.predict(new_xv_test)
    pred_GB = GB.predict(new_xv_test)
    pred_RF = RF.predict(new_xv_test)

    print("\nFake News Detection Results")
    print("---------------------------")
    print("Logistic Regression:", output_label(pred_LR[0]))
    print("Decision Tree:", output_label(pred_DT[0]))
    print("Gradient Boosting:", output_label(pred_GB[0]))
    print("Random Forest:", output_label(pred_RF[0]))
```
